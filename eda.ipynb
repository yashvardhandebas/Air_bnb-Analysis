```python
# 1. minimum nights < 0
neg_min_nights = df[df['minimum nights'] < 0]
print("minimum nights < 0:", len(neg_min_nights))

# 2. minimum nights > 365
high_min_nights = df[df['minimum nights'] > 365]
print("minimum nights > 365:", len(high_min_nights))
print(high_min_nights[['room type', 'minimum nights', 'availability 365']].describe())

# 3. availability 365 < 0
neg_avail = df[df['availability 365'] < 0]
print("availability 365 < 0:", len(neg_avail))
print(neg_avail['room type'].value_counts())
print(neg_avail['calculated host listings count'].describe())

# 4. availability 365 > 365 — the big one, investigate properly
high_avail = df[df['availability 365'] > 365]
print("availability 365 > 365:", len(high_avail))
print(high_avail[['room type', 'calculated host listings count', 'availability 365', 'minimum nights']].describe())
print(high_avail['room type'].value_counts())

# does this cluster on specific hosts? (bulk upload error hypothesis)
print(high_avail['host id'].value_counts().head(10))
```
